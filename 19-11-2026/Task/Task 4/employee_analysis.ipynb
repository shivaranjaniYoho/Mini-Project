```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load employee data
df = pd.read_csv("data/employees.csv")

# Display all employee records
print(df)

# Display first five records
print(df.head())

# Check rows and columns
print("Dataset size:", df.shape)

# Check missing values
print(df.isnull().sum())
```

```python
# Count employees in each department
print(df["department"].value_counts())
```

```python
# Calculate average salary
average_salary = df["salary"].mean()

print("Average employee salary:", average_salary)
```

```python
# Count employees in each department
department_count = df["department"].value_counts()

# Create bar chart
plt.figure(figsize=(8, 5))
department_count.plot(kind="bar")

# Add chart title and labels
plt.title("Number of Employees in Each Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")

# Display chart
plt.show()
```

```python
plt.figure(figsize=(8, 5))
department_count.plot(kind="bar")

plt.title("Number of Employees in Each Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")

plt.savefig("charts/department_chart.png")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(data=df, x="department", y="salary")

plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Salary")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["salary"], bins=5)

plt.title("Employee Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Number of Employees")

plt.show()
```

```python
# Chart 1: Employees by department
plt.figure(figsize=(8, 5))
df["department"].value_counts().plot(kind="bar")
plt.title("Employees by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.savefig("charts/department_chart.png")
plt.show()


# Chart 2: Average salary by department
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x="department", y="salary")
plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.savefig("charts/salary_chart.png")
plt.show()


# Chart 3: Salary distribution
plt.figure(figsize=(8, 5))
plt.hist(df["salary"], bins=5)
plt.title("Employee Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Number of Employees")
plt.savefig("charts/salary_distribution.png")
plt.show()
```
