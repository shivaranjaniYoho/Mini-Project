```python

import pandas as pd
import numpy as np

# Read the employee data
df = pd.read_csv("data/clean_data.csv")

# Display first five rows
df.head()
```

## 2. Department and City Analysis

```python

# Find average salary and employee count by department

department_analysis = df.groupby("department").agg(
    average_salary=("salary", "mean"),
    total_salary=("salary", "sum"),
    employee_count=("salary", "count")
)

print(department_analysis)
```

```python

# Find employee count and average salary by city

city_analysis = df.groupby("city").agg(
    employee_count=("id", "count"),
    average_salary=("salary", "mean")
)

print(city_analysis)

```

```python

# Make all city names consistent

df["city"] = df["city"].str.strip().str.title()

# Check the city names
print(df["city"].value_counts())
```

```python

city_analysis = df.groupby("city").agg(
    employee_count=("id", "count"),
    average_salary=("salary", "mean")
)

print(city_analysis)
```

## 3. Correlation Analysis

```python

# Convert joining date into date format

df["joining_date"] = pd.to_datetime(df["joining_date"])

# Create joining year and month columns

df["joining_year"] = df["joining_date"].dt.year
df["joining_month"] = df["joining_date"].dt.month

print(df[["joining_date", "joining_year", "joining_month"]].head())
```

```python

# Select numeric columns

numeric_data = df[["salary", "joining_year", "joining_month"]]

# Find correlation

correlation = numeric_data.corr()

print(correlation)
```

```python

# Get correlation values

correlation_values = correlation.unstack()

# Remove duplicate relationships

correlation_values = correlation_values[
    correlation_values.index.get_level_values(0)
    < correlation_values.index.get_level_values(1)
]

# Sort and display top 3

top_3 = correlation_values.reindex(
    correlation_values.abs().sort_values(ascending=False).index
).head(3)

print("Top 3 Relationships:")
print(top_3)
```

```python

# Select numeric columns for correlation

numeric_data = df[["id", "salary", "joining_month"]]

# Calculate correlation

correlation = numeric_data.corr()

print("Correlation Table:")
print(correlation)
```

```python

# Get correlation values

correlation_values = correlation.unstack()

# Remove duplicate relationships

correlation_values = correlation_values[
    correlation_values.index.get_level_values(0)
    < correlation_values.index.get_level_values(1)
]

# Sort by correlation strength

top_3 = correlation_values.reindex(
    correlation_values.abs().sort_values(ascending=False).index
).head(3)

print("Top 3 Relationships:")
print(top_3)
```

Observation:
Salary has a weak positive relationship with joining month.
The correlation value is 0.190131.
This means there is no strong linear relationship between salary and joining month.

## 4. Business Questions and Answers

```python
df["salary"] = pd.to_numeric(df["salary"], errors="coerce")

print(df["salary"].dtype)
```

```python

# Question 1: Which department has the highest average salary?

average_salary = df.groupby("department")["salary"].mean()

print("1. Department with highest average salary:")
print(average_salary.idxmax())
print("Average salary:", average_salary.max())


# Question 2: Which city has the most employees?

city_count = df["city"].value_counts()

print("\n2. City with most employees:")
print(city_count.idxmax())
print("Number of employees:", city_count.max())


# Question 3: Who is the highest-paid employee?

highest_salary = df.loc[df["salary"].idxmax()]

print("\n3. Highest-paid employee:")
print("Name:", highest_salary["name"])
print("Salary:", highest_salary["salary"])


# Question 4: How many employees are in each department?

print("\n4. Employee count by department:")
print(df["department"].value_counts())


# Question 5: Which department has the highest total salary?

total_salary = df.groupby("department")["salary"].sum()

print("\n5. Department with highest total salary:")
print(total_salary.idxmax())
print("Total salary:", total_salary.max())
```
