```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the employee dataset
df = pd.read_csv("data/employees.csv", sep="\t")

# Display first five rows
print(df.head())

# Check columns
print(df.columns)

# Check data types
print(df.info())
```

```python
print(df.columns)
print(df.info())
```

```python
print(df.head(10))

print("\nMissing values:")
print(df.isnull().sum())

print("\nDataset shape:")
print(df.shape)
```

```python
# Count employees in each department
department_count = df["department"].value_counts()

# Create bar chart
plt.figure(figsize=(8, 5))
department_count.plot(kind="bar", color="skyblue")

# Add title and labels
plt.title("Number of Employees in Each Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")

# Display chart
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Create histogram
plt.figure(figsize=(8, 5))

plt.hist(df["salary"], bins=5, color="orange", edgecolor="black")

# Add title and labels
plt.title("Employee Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Number of Employees")

# Display chart
plt.tight_layout()
plt.show()
```

```python
# Count employees in each city
city_count = df["city"].value_counts()

# Create pie chart
plt.figure(figsize=(7, 7))

plt.pie(
    city_count,
    labels=city_count.index,
    autopct="%1.1f%%",
    startangle=90
)

# Add title
plt.title("Employee Distribution by City")

# Display chart
plt.show()
```

```python
# Create box plot
plt.figure(figsize=(8, 5))

plt.boxplot(df["salary"])

# Add title and labels
plt.title("Employee Salary Outliers")
plt.ylabel("Salary")

# Display chart
plt.show()
```

```python
# Create scatter plot
plt.figure(figsize=(8, 5))

plt.scatter(df["id"], df["salary"], color="green")

# Add title and labels
plt.title("Employee ID vs Salary")
plt.xlabel("Employee ID")
plt.ylabel("Salary")

# Display chart
plt.tight_layout()
plt.show()
```
